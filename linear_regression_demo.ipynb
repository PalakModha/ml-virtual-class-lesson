### Machine Learning Workflow

**Data → Pattern → Model → Training → New Input → Prediction**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("Libraries imported successfully!")
```

```python
# Create the educational dataset

data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Exam_Marks": [35, 42, 50, 55, 65, 70, 78, 85]
})

data
```

```python
# Save the dataset as a CSV file

data.to_csv("dataset.csv", index=False)

print("dataset.csv created successfully!")
```

## Understanding the Dataset

In this example:

- **Study Hours** is the input feature.
- **Exam Marks** is the target variable.

The model will learn the relationship between study hours and exam marks.

### Feature

A feature is an input variable used by a Machine Learning model.

### Target

A target is the output variable that the model attempts to predict.

```python
# Separate the feature and target

X = data[["Study_Hours"]]
y = data["Exam_Marks"]

print("Feature (X):")
print(X)

print("\nTarget (y):")
print(y)
```

## Visualizing the Data

Before training the Machine Learning model, we can visualize the relationship between study hours and exam marks.

Each point represents one observation from the dataset.

The visualization helps us identify whether there appears to be a relationship between the two variables.

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    data["Study_Hours"],
    data["Exam_Marks"]
)

plt.xlabel("Study Hours")
plt.ylabel("Exam Marks")
plt.title("Study Hours vs Exam Marks")

plt.grid(True)
plt.show()
```

```python
# Create the Linear Regression model

model = LinearRegression()

print("Linear Regression model created successfully!")
```

## Training the Model

Training means allowing the Machine Learning algorithm to learn from the available examples.

The model observes the relationship between:

- Study Hours
- Exam Marks

and calculates a mathematical relationship that can be used for prediction.

```python
# Train the model using the dataset

model.fit(X, y)

print("Model training completed successfully!")
```

## Understanding the Linear Regression Equation

Linear Regression represents the learned relationship using the equation:

**y = mx + b**

where:

- **y** = predicted output
- **x** = input feature
- **m** = slope/coefficient
- **b** = intercept

The model learns the values of the coefficient and intercept from the training data.

```python
print("Coefficient (slope):", model.coef_[0])
print("Intercept:", model.intercept_)
```

## Making a Prediction

The model has now been trained.

We will provide a new input:

### Study Hours = 6.5

The model will use the relationship learned from the dataset to estimate the expected examination marks.

```python
# Predict exam marks for a student who studies for 6.5 hours

new_study_hours = np.array([[6.5]])

prediction = model.predict(new_study_hours)

print(f"Study Hours: {new_study_hours[0][0]}")
print(f"Predicted Exam Marks: {prediction[0]:.2f}")
```

## Visualizing the Regression Line

The regression line represents the relationship learned by the Machine Learning model.

The graph shows:

- The original observations as points.
- The regression line learned by the model.
- The prediction for 6.5 study hours.

```python
plt.figure(figsize=(9, 6))

# Actual data points
plt.scatter(
    data["Study_Hours"],
    data["Exam_Marks"],
    label="Actual Data"
)

# Regression line
plt.plot(
    data["Study_Hours"],
    model.predict(X),
    label="Regression Line"
)

# Prediction point
plt.scatter(
    [6.5],
    prediction,
    marker="X",
    s=150,
    label="Prediction (6.5 Hours)"
)

plt.xlabel("Study Hours")
plt.ylabel("Exam Marks")
plt.title("Linear Regression: Study Hours vs Exam Marks")

plt.legend()
plt.grid(True)

plt.show()
```

## Model Evaluation

A Machine Learning model should be evaluated to understand how closely its predictions match the available observations.

For this educational demonstration, we calculate:

- Mean Absolute Error (MAE)
- Mean Squared Error (MSE)
- R² Score

These metrics provide different ways of understanding model performance.

```python
# Generate predictions for the existing data

predicted_marks = model.predict(X)

# Calculate evaluation metrics
mae = mean_absolute_error(y, predicted_marks)
mse = mean_squared_error(y, predicted_marks)
r2 = r2_score(y, predicted_marks)

print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"R² Score: {r2:.4f}")
```
